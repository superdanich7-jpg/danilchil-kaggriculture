## An inspectable and editable livestock policy

This recovery revision restores the V232Late feed baseline after the bounded double-Yarn annex returned two completed official samples of 600.0 and 696.1. The annex is retained in prior notebook versions and local evidence; it is not the default policy in this revision.

`AGENT_SOURCE` and `main.py` remain fully editable. The package exports exactly one deterministic root `main.py`; local screens and Kaggle's completed submission rows are kept separate.

```python
# EDIT HERE: change PLANS or ROUTE_MAP before rebuilding.
# To edit the complete policy as a file: first run this cell once, edit main.py,
# then uncomment the following line and rerun this cell and the demonstrations.
# AGENT_SOURCE = Path('main.py').read_text()
# Example inspection (no change): print(PLANS[3][144])
# To load your edited JSON file: PLANS = json.loads(Path('actions.json').read_text())

def rebuild_agent(plans, route_map):
    assert len(plans)==13 and all(len(p)==719 for p in plans)
    assert all(isinstance(k,tuple) and len(k)==2 and isinstance(v,int) and 0<=v<13 for k,v in route_map.items())
    payload=base64.b85encode(lzma.compress(json.dumps(plans,separators=(',',':')).encode())).decode()
    replacements={'_INLINE_TAPES': 'json.loads(lzma.decompress(base64.b85decode('+repr(payload)+')))',
                  'SHOP_PLANS': repr(route_map)}
    lines=AGENT_SOURCE.splitlines(keepends=True)
    nodes=[n for n in ast.parse(AGENT_SOURCE).body if isinstance(n,ast.Assign) and len(n.targets)==1 and isinstance(n.targets[0],ast.Name) and n.targets[0].id in replacements]
    assert len(nodes)==2
    for n in sorted(nodes,key=lambda n:n.lineno,reverse=True):
        name=n.targets[0].id
        lines[n.lineno-1:n.end_lineno]=[name+' = '+replacements[name]+'\n']
    # Preserve exact published bytes when the decoded inputs were not edited.
    unchanged=plans==namespace['_INLINE_TAPES'] and route_map==namespace['SHOP_PLANS']
    source=AGENT_SOURCE if unchanged else ''.join(lines)
    data=source.encode('utf-8')
    assert len(data)<100000, 'Source exceeds the competition byte limit'
    compile(data,'main.py','exec')
    Path('main.py').write_bytes(data)
    return data,hashlib.sha256(data).hexdigest()

candidate_bytes,CANDIDATE_SHA=rebuild_agent(PLANS,ROUTE_MAP)
IS_PUBLISHED_BASELINE=CANDIDATE_SHA==ORIGINAL_CANDIDATE_SHA
print('Source bytes:',len(candidate_bytes),'SHA256:',CANDIDATE_SHA)
print('Unmodified published baseline:',IS_PUBLISHED_BASELINE)
```

```python
public=evidence[evidence.panel=='Selected public histories']
paired=public.pivot(index=['opponent_label','seed','seat'],columns='candidate_label',values='margin')
delta=(paired.V224C-paired.V221B).sort_values()
fig,axes=plt.subplots(1,2,figsize=(13,4))
axes[0].plot(sorted(paired.V221B),label='V221B parent',color='#6c737b')
axes[0].plot(sorted(paired.V224C),label='V224C timing',color='#256b78')
axes[0].set(title='Public-history margins: central range',ylim=(-12000,16000),xlabel='Cases sorted separately',ylabel='Final cash difference')
axes[0].legend()
axes[1].bar(range(len(delta)),delta,color=['#b75b4b' if x<0 else '#256b78' for x in delta])
axes[1].set(title='Paired change in margin: every case',xlabel='Cases sorted by change',ylabel='V224C minus parent')
for ax in axes:
    ax.axhline(0,color='#888888',linewidth=.8)
    ax.spines[['top','right']].set_visible(False)
fig.tight_layout();plt.show()
print('The left axis is cropped for readability; complete values, including extreme margins, are in the CSV.')
```

## 3. Why selling order matters

The executable market-flow reconstruction below preserves the earlier diagnosis: shared orders execute sequentially, so timing can help or hurt depending on rival actions and town consumption.

```python
FEED_AGENT_SOURCE=json.loads(lzma.decompress(base64.b64decode('')))
import zlib, copy
RUN_HISTORY=False  # 636 historical games: feed baseline and timing baseline
RUN_ABLATIONS=False  # 48 full games, four candidates
RUN_REACTING=False  # 96 full games, both candidates
EXTRA_SOURCES=json.loads(lzma.decompress(base64.b64decode('')))
REACTING_FILES=json.loads(lzma.decompress(base64.b64decode('')))
EXTRA_SOURCES['feed_baseline.py']=FEED_AGENT_SOURCE
for name,text in {**EXTRA_SOURCES,**REACTING_FILES}.items():
    path=Path(name);path.parent.mkdir(parents=True,exist_ok=True);path.write_bytes(text.encode())
FIXTURE_DATA=json.loads(lzma.decompress(base64.b64decode('')))
for name,record in FIXTURE_DATA.items():
    payload=base64.b85encode(zlib.compress(json.dumps(record['actions'],separators=(',',':')).encode(),9)).decode()
    data=record['template'].replace('@@ACTIONS@@',repr(payload)).encode()
    assert hashlib.sha256(data).hexdigest()==record['sha256']
    Path(name+'.py').write_bytes(data)
OPPONENT_HASHES={name:hashlib.sha256(text.encode()).hexdigest() for name,text in REACTING_FILES.items()}
OPPONENT_HASHES.update({name+'.py':record['sha256'] for name,record in FIXTURE_DATA.items()})
CURRENT_OPPONENTS=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4FPFMc1dAD2IiMdjpMse1gX7kgFIzP4U+Fpkf+XeY24x3faPF3DmleURNt72x3A7a3PumsFEJ5xPE6ZVYBz/Cc/hpaHBLJXPGX3igc2KA9kv6AK+2+vyVk1z7r08/0EUey2w3Wgy+zkuTM4bKl1jZCx0zlkG5BfQ/RZ93f2NRtShv7w5K7huUt5hS1IKmnU4VGOvyia3p08Kx7EN+HPrPNovAPXE2DTn2Alts6439WWw6ptiRtcRavSoEZgWc9lmGLp8swfqfjRkPkkRzLKfsV3+KTl1yUNO0HzC1/+E8/o4HuQhNQMyQg7LHKYauWo1nr/C7oH6E8fMVzQZENrImzQxW3HUikpNkfPgHT80ah4IJC0ytarr1vwqSQFgVntwMNI0MCqI7L1YPSHUCXTt2Yuq9Q/yFQ6LxBOw7YUNaehrEDq+7ckvXQ5JcbxkWGj9+oYa4BmEtosP59f0O9OEQJ571pfBmPhC7xZamPQgValUAHaIX0i2szUT8nZo3n7CRySMmKf0+TLOthYX2Cym9Cpl3v26j4lJSBBkVlvFNSiP/5ZMv8PY2vtXQK9wx9NJz3zicZdZM0+oJyuff27/gtGFtMnGnUibRw0qGBbO0YFCFmVeV6H7TZdlz54dGK4qaIbBMQ9p13RpmubykuNjSslP344WqdOUf5frup0hLZnEj9TjcUmLLOj3oHhjNkkGVroOm1CN2u8uJGDCQm9CiToEkYCAyA66NxPPoFcN2qITXTTfiYASnpPFv6mva7rTAEIOSVsdmVYl8o9wmJh4NVxdNSyxGc0QLvYqVks9jdYOsOWO+iS+WnHmQcm76XbTI2A5L9+qJjN24G1nQcJBEg3tsQ3jQeuPN4dF9wHq95ROJ81bt5K+hKsutHjii5Z3TeKyqMDFo33fDRTix10tvMmBCEk3Hu9xp+jS49/8avBIOdxn7I40k6zcbE8wbogiEgRK1mNxUkaV65SVH8Q3QHaykAwye0nr+klTe/3cxm5OFkT2nVZaaU10f6NCE7Ls2p7jj+FvXYCgkJoz2ooFN+3Ac1E9BH1iZCvhIrqGNJrVNXUHbJZON8wxFw0gbfpoiSBK0fwkRKZv8VYpRbPj5604sBzPfTfXEM7xI5V5eZB6wGPT82uITYjThVNlsSZtgdvQSW3xHz3uww7D9LFdys0WYA/dfbZ7usASXfSPf/lLYIeHxpdMOJktg/6Qnmq2POIriO01PplmuKAoexH4M+93DPg79GLOWtrFhao6Eb5oKeS74sf9pmpd2R09jmBoXFoY4WnR5Q4J80URENG+OFhFrGOnZtaPE7F+SavdXN81tjBtVLPGdxOCf1/OQ5MRSpJ6fTwKOvLn1J0JSv41wOHoIxFIlZzsM2CkohUGguvKw7norEgG8EVC4oHgQMUiTzSCYl/fyPJrlJMK7vdlMlx/PGO3NjDqoe8OcXgb1i2mua6Suo/pR49CocH1lSpkWIq03yHILAN/MRLUOPZLwiyEZyj0YikSk+QaCeQ4AaBT5/EFdn7B68FVdvB7ectmGY4E7bcp0nbsG+BhHQnfENNQ46fvxHgfZxy7wjEd83zmukDdgeMEyZr+/YDsxZD9DbNMTUa+jeK5uzf6X+Jxpfh5BcTNHrqx2b1eVlrnRIF9AvOdINoBN2cJLg1jL2WckvZ9EccULXLtbgB/BO3aXAo5BtqFxD18l9RoE07n/gkvzrP2aU50Q/mT1wekaUwQZHyKpZJaH4975F+OROKrfMid6NyFiWFt40GH9tzAZWF+LvtzHDo6RzW6OnsRpJA0wjp88wBc4y/L7j7M0aKUevo6++rLQ1dIT2Uo2qCfkKFN/O8yu2LCGJAw8JBD8y9PeoICBGnfvzb21LQrnYzwc4tns4K4CAPHVvXwC0covvKdpc0rtDfqJGOAw8cr5RYaJ9gqeb6sL8Isx2BS4Dy1AEzje3ZQrdERKV9ON5wqGHVOOk3GOq2pkWnD7EjWGKM0CnaPAJPiCtQKqSEUlt0FyFdzusS39YZSI5svFopzAzfW7iHVanaB0nJ6kru3ZBjuVfdRxF6p0AN3oI409Pw/EwYWyTagH84oKR3N8GOSYN/ZidaSkxWen5QApBhYZ1BVYFk3jY0QTUCeSyRje6bqOnGkUSdULn/SwKXGL/RMh6Hz9mN7R1fYb9MIVHs1uDMLcIz+COaGn8Vc0IoCo2+bQm/42g8JXeg3l6DQapn1nVWl4P+s8AYszIgyi8BdHBB1efIZp+g22FAR/Sp90JS9t8f2uj2oyVu61Fb0x4HR7qIsQR7ZOAqDnpp9nl8QKTHrIOJUKjvhAxLrGmrH5+ZKoiWlmeLPsyrH9mkqPxemnMbp3i+EHdPutQjcYSJW5fe3jXxU8WkET7cDNHUN1+xPC8cQH7/5G3GghvK+JqCuz23xssP+qnTwi85f9+pNajhHouLuC4aV0XsVU1onT4ERDejO3rIbPew12j55VCB4EC8hR1OCJNm3EPY3ORKNE425SKtrcpwaTqaUemtC2/T9QK8bQy1uiSu9f5XQRtFZt+d3a4Obd3PfZrUaK25BCELSpAtC5+3QvE37QH4qCHiJ0GEoqeQETSNNcHMoXJgCHXOyZzxuRiAVQrMoys4Q52AaFGWyGPTgNphZJ22bQ4tVrDzHnvd/csZP5593Tx6TFuufjjnpz49HXJ8Z4aC6p6FElAByuY+je35c4TD9Q5LQc6im4swcNEwzgJb2SpGVCJXeLGflKykJiuPDpg+EdWi3ooGAHG8YxhQIKTBpqRCUdAfiOyvSXVRE+63KmNo0AcY1dM1mMApnmPLNitqtV1R7mbfWnbiTT2TSEtj3JNB9v2xUvCFlFfvwd5iUbpvcQWGYaDT9DD20dinZHb6eUfhg09HSGSG0+Jb/KGfJZoqL4veRlnt8Fl3PQ1xSK1b64OuaXWssfJH3lhQ4i0wKedqwIUZetGtxVx8XNwVNmUz8QMZVBi3qoTjYb2AXTWAFZ2kJwqvrcvkQimBptl2Z9ahfC+k0eK6VqXFCzWDiEBtqMCXT9aiBfK2fyiuvIBgE2y05fgwcwb2+MqE8uIRxj1rljUBu2+eSjtSoGuMASsOQ1jNolGScacleWClM+DcetlA4zHgrKnlt8wSbHye7JPAvSNW7Lai2FSSXltjZ2ICb6KCckc1/NKBpnIL+5BVmR6BhgRPaksry7kW/he4QzD91V0eHHGN5lssBSO0NpJGQq0dN0b8V+X8AabHJveXuNEg9ZDh7QcOrKN9mLEzjSdfBUWDcHMqhdQ4RunuoAKAkeJzDZ5yhsiSm6AmEwr7/Lf7DNMRNMmF+wiWrEmCT4JScjqu5kP5pBVSBVSWNDmNpsRp5pyTM9STy0d1oVc6OAhswRd3Xej/xQGwuGcOnbMrbdPhfSdAvyRDV4/yAvYdKXQ2vNqCYg1kRsdDEgFedSfJFnSp6Dki+d8KKfngo2REsyfotsVVfVe8GmX3v0NFp5akPkYlSL9IcqKHJ2Vmt62+peUnHnz5JeNws2hcHeENu9ViytRhJCHQ7B+ESmvmIZMLPhLNZHpItNfHg+B33eAAXDfuD7nwY9AaGvEYbSJhz2VWMGxdKtf50PbK9HUh3WtL4MMy4NjJdW9UCQFcwFGOou7WgDhg77wSm9WPXRIIejLhJxP14OAIjw6MoU6tztGGzFUTbb7ILaIwkXFk4ojwE22M5qik5AgeK33yWAkHOn8bpZqxkH5nKJw+T0upMS0XqIsj7SaH1EHdEQGKTjTD9F7TVVoL6zFzzSRzCKtR2ewb5itnxh5W59pq5UFZcVAohewOteXOODp4kNiTwhxjQVDQq3KvQMo/5HTBSp/OjrMyybKh5Ne1vo0RAz/btaYXtJ6gotnk3TEptPgmQUiYNyoHSIyVQ5/bmXCFrFDd5W8Ml/zoyHSXJl8UaKLwl+PTRdBRGi9CFOIK8mvUkJbd8W7Vene2qhARGMBZTEfi8yGINBUebcowRlGyCJtk1wWcwZtwXxMIdrxr5f4ClqaQ44rnT+eRxrsQV9SAC4Jg0Cr96HjE6kUOfpTwPFK9BrFNtzg8qf0+1h92Yiwad0Ga7+TTgp+FtdnOFWwwS0S6vmT6evVmU2UgUJ+YD1y8UksUY8yDt4XggaEG0rQI8cNmyQGVAtGDbV3XE6EqhqUjTTKimt3S0jqayW6/VZEmIy26gbvv8ol6l1cseVoROQhS7vCaubME1RYjXtzXUqcVpyicZqwXZf3plzrEaT8T2fiPxtiK1VYGV6gkL63ifvRc5cGpT7Cm2nANxymfabKKnvAoIXaam9PFIbIuD1BTt3fLrh3S0SoEq1ZbgEAzkpHUPNFpBio1UCBX+WbhmJCwaPvEKCyet1c2oWUFV/RsxqxAjUcxmk4M4UJ64F6e8iQYhAjS0MsR4dKZqhPQjQl+NVUySJIEYAqKgnV8v0Q+MZiTkQcPBPHMCo6h9GGXj33O11m0337s5PIriptPH00ItCGbi3ynCwutg6U47gUy7g5484G01oH3TjSiM6F+8tWR3HQU5Dg3B9ABmizS27l1wCebZCyeHEwgq5z4zfrA0+fk2mUp7/43cb27+C/eT9c7bouNc0/v3dwyP4p1DLaQV6UBk2c4OJv8rYHosD4K32m3772q5I9Rz8/tna1HU8KuHLdsTGu1bZInUTqVaY1oFrIrXXdkcn/2A/A8yWnXz4YJGh4aM7UmUgE2wmjdTzS536t3Bj3zdGfAcC7wYvngewtVnYW2E7QPHGIm7j9F4xSOdIH0aY6Nn4UT02D6pI+gYyHIokDGmijQjcjHWqnuFjefePTH5SJhXJ1KQiVP7j6gnBFDSmXgtzTobyaP+YnfzgC3syHp4OHVeJ4e9XN0Euc2C1QeJ2RpCB8zoUo/HnbdsfPzojAt6HJrXMMoH4IIvjMpgsWZMcBOkF3Ubq74N+hnxgjGpnR261mzoZtopTj7rDWgBGHMdwfQ4WIoPCJM6y6qVbhShqkFKzGdsSv+AuK1TBhscmWObxftd8eUujLYrNTSucaSIchDHShovA+Wv/kRpHLtMtBMKpCxev8IHCQ5bi1CMeguVF8EhRAZOEP88IQ+AW9UhV6wBFXRp6AyPoB7TlltCoAabSyJ6sOjZmFCu6R5QzJpzGRKqPeRFX5OpDrBannNupOV3XfRHSgXgVkvisvuSK9F16lsmf90af5+sTQ+subtdRLWeZoV+KpWI9nVlz6kQgWANvTlH/sWQAc4QZJoARtIkQbI9FgnMluhcCOyBIThbLz3ssovJGZ7GjZhZ9BVmUKOOy8njU/TX3iq4Y7+SioDTrg0rMt4IdgnS0r5lgyQUnY1VZTBPik9FJxxqTU+VDg8bORvj0ZBJ3X89xRfgGxaeySnf8O1cOg3DAgwFsDs7vmEmTkIYX4uQWiGcTnjNISKlZsL0TLaUTwfQPvjX4axRtohSDZ7dqErtsE7fo5J2Rv+WFB6uX2w1EQg6M5Kj/kuToHREzKMJSd4LEJFeWlbXYFe7fhV8tsL1yhEBdXbg+Chtg6RWVhJdtysyIkgFd2TkOxS+16vdWHuq9F5Nnu+1hsr4dzcCnCLfhKrnIChyTDLFeJVUSZAEWA4U0s1mg+2KJ8JP1EGN6dUbApxyqazofqPwTVGWraIkvUvMdvANeZ4eKKH2c/F313Gx8Ne6Ul186LrJS+uj4B9+AWBxnoYxPYWAYgzkB6jbSAv+a9GbCpqIWPoylpsZrPjJQYDQZZosw0QLJ/0s6rnJU9bMz80OASthW7RnTPLI/gB1yJKMQuu6y8cjEyzZ+PwHl4oVXYtnXWoZ+90WdhoKVD+yloXWqAXHed0pIhycdRNUMmGx22ZXzvNO8nqCr8SeSJ/osXmJ3o/2mVAlyLUabKIg4Jc4hjTfICYN4hOfVbFY01M1ZUrxB3p3tBDrkfxRwsREvPB3atJHhhTkqrdMXDv32hk3r0AeKuYaKOdqUDyKOzeoi42fBGHAn9DrighKrKAHlMbAQp1JpA7uNY63wlz4kBvsFDYbiKhcZQygwZ/pukGeZ/NZkQL62GdqRQHuauK72vwPBbwtqYTknkYmCIAgrJmH7zyoiFUeyt6Y/iYB77k4VEOBJz4s0YQY32eQcohZ5iON6aG30hpAOwc+sO2N2KP7hGiONB5O2fvYoAW0ig3hjqzoWewpYePXV9AVRF4DGC92TnazOyVG4jI0Qnr4iV8O0GF3GYuvmQqlxGsJDCJ+lQkyZyjmOrzMPorlfx3RoDN93wvuBtB8Z0xSMklPOna+BpiSyz84sG1U+WE71urPcwnoywVpE4c0dfwy5UTlMHwFyumizIwbYrGZl91rZEojB1kvzjnl6czJkL87ZXpMxR2KO8kDlWlFKfS1PlaNHwkFm2v4DF/BRV6zTA2Vt/z75lHTnnF72cb0+l9AWOZ7YFmFYqADpB9bcakIEJGlL13u5dhNisadTdrmORKQ/k9y/LCrw9pswQbrx+h//F4Udnw+EkSmDVSuf6d2C+lXLYmme5drazFXLUeXn7z5AKnDBSF9X7yHIgj9akx4tvXJro5FhSPSYI81ytellRA3K8ftTt+kbAnIfQH9y4ifwuSUp3YqBhoNl+InPoeKsInApS960isC/YtQApV58XehovorCciGWbiDzz0UCQDpQ+HNwuymTD8fpDt85f4ix8Gd7OU023UD/gzgPwiL+fV8l/f3pMXcClN8zG50HD0bZBGpF2uMAae+aV9fqgl/3YAEkSZ4EZ3nCkoJ/lveYfZ3735UYMZPAEOmLAsI7huh/UnD+OIYm16+KB8nYD9AEH7s5uS+C2FOFsXyGQIvIZgFpS235R3kaJ/k8W7REHsvcpMWOmiFBmF9jVbNIN55nuu0uZ1UbH0UbMOU3ir0D+YuTAinblWB4U675hwYqrFwt0u4G4SmLQuK6OXfEfEr+a783gqy/HPRic5mTQcPooy1GsVJ0KBjKfelPUO50DeGMn9/bxdr1qvYBPEN58z5rVrXlundNi5HftqC3tLoAGn/NQjeVWCs5g/4JFHxP4qX//NQV8k6B8GCqa9Ss4zqcDiGlX7hgDBshcLWAPf5yhxTv+CXgYrvevMG0EZ+JZOntjeXvmP5XsaADH7f+XZ+FBsKjJ/t5dLxRPgku+lA9rbFh6zsXZ2kI01XsVqHbE6/Wzx2igFFjfsIpI1ghyrgKqlBsCNb2/oSGNdIW6e4lmp1IiwISKtCicAjSeIc0O5t2GNywncxv/pzCBf+Y9XcMZlA+IYSyrduL9qaeUGq4ZbDZrlp1ysqdcMyzqMB/iUMR4WXMa7Ob2qhsBrQFcYJ0z9n+I7soFbhYekrTrHG0MEC8uVJfxmJ+Xz6qsPO+3cTWo99y3PgwhTmpglOzeTOMGNo0qurtbqMbKYL/Oeuzzoln22XKT1rFp1mZAwpl2Ntg29Gj/Y++pGxpNMgKfHALYStWii8nuPcFH3zHzqYcqLEa+muaaZ/fqwHI4YKSHJLOXbHoGWMHx1qVQfHSSDq4bPoLhYyUWnFAd3EAruf6EQVbiaCJPpVUkod3PGMjM8ICtXXBuPIKJJvm7CqgC0/FXngh2pEi8SvDDxwP21GaCjvoRidwXH3GOIkyGKjeUS9KtojPGrs1RH5V/UKxNcinhsvMtnfn2t++2dmoaP3rdZn3iRM/HG8XMgOdNywbdooJGt+wvCk8IxLoOvBahF90noS4jggqGmJzEAgKtoCSQC9NKswqfEDHrWWNLWDDak4AYvwCypSObN2X+2Os6ApGVg67gBOhbVH4YZzdQ3L2tVAUay5s/aCWr5UjOMatLErN1s/lJxhjQ2Gf/vzgrJZWwY8vP21AMRV2veooWuQW46XA7xS2Ugtb3DMKzXQgtyjIUVM+q9qSYU9bfD+401m2LUlk2m2LNmk/Lr9P3Cj7hd3uCq8tE8y6GxxHW+xHt2rC34ufa4Y1eNAUJn2JYNRAOh3Y1l6jtByarHyixqRZQVfuGjG+177xBKKGfPydZA9F08Bnh/dSNHz/c8V16Ak6m4Zbhya+JsAVU0vc3m6s11RafH1q6HZWQZe2NIvDvRcBLfbQ3lJiCLUx7jfcTxrNfjKkSr9458gT1K8K8rf53JodeUPvXJDreWWOYphreIYU2/49AOiaXHXKDRgeKwmg1fFi+Bzbekeic3o/b4wPJFVluHR/0338nfTo2KDUR86LXwApJINzCDGHdgvZDn48xJPZ/MbUwBb6CBOBjnaEl6jcZJ+R3C6e29QaY9Ho8jtfDdkAwTaeeeeMRRhJSWdMO/fxxGkbgSjc7K2KJ18ZKEbRhGetbyhaXRuNzIiXsHBrjDHgW6hhzVI6Z9UyqwAlsTaKOhQtlSrZfuiuRtfSondbEI/BgNIyz3+x8KpmK8N4+aKVAxKufKuv0FL2Rq/u1uOs22ptaF8m1T03Qs8p7OUyWaANC2oX2KYjL797CPj07JvWftOEbTq6toyhT1QFUHbvCvBVdHg1iLAIhO3T+I5WRf72TcmJSsalIPriFIPEJAoaPyjjCH2sCHhpXeG3nsWAm2lsAXaMdmUyeAJpnScss0+kciNAXmDnQSgl6LL1RDXEWe5jlwswTqq4z+dPYycWpxE/MffjXTLFlVCS8RzH8k/YjK4ozJWk2XwtnPWDOy1XJK12ytOoiK9LyeiNlQDeUMDHLmX6PCqzYTHE3c+MudZ25XPTMGGeFllUS3rjvjaYCMo20zu4mOYdbv21H+JED9RoggYVouZCv6smbW/zcURnMqjacZoBKD689KVxWuT+GfDKzhrpP4faR71Mx1x1UpiFEu3154kkuU6109l8FkPj7BhqUK0jgUPapnGPONFly9Cqnnht041TBZzhbwv7nMIHMaqDwan/NvoBQ6KhxhZOGzd8o4zD++4jkZKEEBVXqvugPaIsZBvV/k/w/tKJbT7tZglegJAUTKYJpMCSkQqOorN2FuiRU+bP8cjU8eih/AY3FDngfXMFGQ1MuHC0mLr7knPV95i7ouF4mbUsLVTaFvKIAc2hZehQkTXe2QKBs3BO0U4cvOBtUHOvTky3A9KE5Vq0AaFzl3FF3ZCkjnwHt36H6BWtGpzRMi8MufYGoVoTw/eyK+oRW1crRjqz4M1k9tmQHn6FOjpHM6NiriL+ZsWM1iOYvAdXnZMksBhPOkOVvlvZjGIrriB+lMx64CRVNt5hZaShi+OUtJsNGnoCaCFjUbP/CXxiSpzhqPK9UCSzUgCTh7FtEHnNQrMHrInTp4qrHdaBraG9ZEGE1w4f5z39HxALSfePj/djPjd2NZvAbcoL0m9KamsvhK1Eml8pJe+9ocM68QdIVy0fJ6zG6gGKKabWDSPoduQYmIrXNhFsC5752idngGQlJUdwKi0QKKvWusKzA4TyIMjjz/OPolY4rYp0qNjpXYNQ0nvS0WDqe+WZmB6OMU3o9NUdi4va1z3xQSNElGm+ywGq/hVtZ9vQbE+QRRdD5ENKFpphHy5EuvI0CimongN/FhjHosWauY74d0uji95nJETTQ4HZ5UUFYhb16sI54rSfIRcgDUBxd0RRZW/4W9FdrBStJ3TuodBpHrDBMDlF+t2y5QGG4tK/WWb4NcecbhdwyHf/jz93lu2aTRqVHrvz+n0k3WbhI1U1uZg4Us9Asqo8xFt+hNSO/w059eWUM1teB+KjzM7AA6h6hTYIMcIpiJrn8fKDa3Ayi1eyc8cyKD8piKsfnFCUCtA1dPtA4A1pTJO4+sT/VxkZ+t0+/6QbiNZeCew4qnI7iPawLE6JY0Nawr4OcRKRNY2TwnmveKA0qPAJYr+WUnW7DohaxQV5yYVm1qHUTNuqGNGNNVi5TVwRP8OK3NGMBnqtOvctxuZoLNjNx5r4nVzPkaslw72xnxazDviyEP0crXyzFWwT3mTHHa7SCKl2wSK/1UNwP5xlWBPhD/oTUkcddNJKVgjN00B5vS9eozQWNJo4ZCfVTUtw1DdxvPHUF65hEkSFBXaB2DPTkfCHdOFKPByjFz+4zUtA83c+NNxO+SGu/ytPIJa+zRaFMrYxm++qGwytNYjp5l3TcIokTC1mdp68hxMswyGOxhdNd3gCsyXttSfuCyG6j7fMYbUfaN73Y6X1AhmMizvV3tX2rDudPGNO/jglXgTJYGvUqltzvojdmkV47iY4rrNeyJTXjHKkmj2V65QT5VFemn0/on6Rm13buSr+qPWA7FCGF6RcMoNhxRdvthceqA6UJG2RZfGIHaT0lM3X3DOShCzqYFcrlhlKZJDgSScsjYa4js7UbkcxSRGohvgrWiZZqvHK7e8X8JQJavK2Ug6YOFthVSJUCDS8aF+47V5UDVXK6yqx4OMNDkA8e5s11J3mIsv2WTlOVw2DjC92U+60xBuRhyN6OrF7z7huVqSxQ4IFsWwXmmKN3Oao3D98AOpn2BFwZNy3p/0XpWFi082TZJjHZyF82jd7ST7MmASM9cElM0/RH/Eh67EusxPyZOyT5SL3djlzL5rwZtmtY/eN9lNsBTy1+8dshVZ4cbZATATJMJM4iO/F6TRz1zpPxizmE2MGQnbXEWaZmDPe83G2TUYiDNu4Gzr51LZ4aWiZS9fjTJdKanuit/TS6ekj7P1iJxYEsGPckr6VcoFGOTIBVnyAtSGey+i9xwCIFWY0q6radMBmFJbsvEmkTchYwET8c/Cg/q3ZqebeRgU37fHKzLMB0a8/qDo1n3vpz/XnodZl7as/dQoDFiGKNT9W6kT9sTVW5hCHJTxaGI59rxfTpCJv1mkQfgJ2zGrrQbIUun6VXZAa/WCLE5ROODim6LeQ9+2n7rP2wlFK8ABW+KQZNOetdb6J4PuPeQ64W+R8lAwv7UIe6LL7CCxK6j2Ahj0BPP1+/RHvnOjNEZZ41FJW9LuygZUe13eRzXAIBBnw7ZgoB1qMXuhCMXPooTxyjxkzWKPRe8yKrn0gPqJhLz8udun14GkraKUPQbO3WRQemYihsg9TX7LIFlun8ugfFeo7K0KjVn+QFSUuVBq8VTnkSaoxsZWtuVnjM72bqh+oZrdVMUXSd7qXUCLTB7UZ0lFG1tuvOTJTEpG+0eSLsSlshe1VEr8eSRmqscNyPnPku0X6XFAFJvMH9GUHmMTd4klL0y6rYBCS9Da6xbIKADxG3W/nun9TSvU6Gr12Y7ZCfGeIpwBeifL7hDNvJMymScB3i9/dyn0zrGzPpCu/zKWiPc/t2QS3GPZkYkZdOviWvewtB6Pp9ZkIuGuzSB6IVm4DWFSzEYQAnXHeni4HglEuXNH2/OBOUdhJLFjrqPRf8dSyiy60WzZ8bvl2lUHb2Zx5D1e6FqZqP+wsk2ISmqiMXP9GGBwwbjoz6CJUoygl/mAQ/ck5bE1KdwBOO1DuNGUXOp7u/Hj7viJUCatHkiM3L9uSebDpqwJL99WRetloVMyz8ZuIPI6b0R/qNvCjAAYweySWfH7b9w4f3xHgLmvKJTBQ01RTSX7X4NwN9nhLXAOkMl4GmpeCTp3Lp7uipw94Q/6oAwjLaKhOC6mGctdRqwaWBtu+5x9uRVYJM31EDpM+zj+o03ZrSmeyE1cq/2af4e+dqbyrkQS/n48fv4my5xa85WVDvns4reMxRfSGsUZYa6GX+TSdJHnXIuH6psYLV1LSwmFBcMaJLAqL0Z4V5TuAGwoty+wwPIqn5o2aKXvW3PH9H2zkRzfoiqvE9ognbGWRMJW7kwCIc11ycEJKBy42Rt16ajHAs3lW3QE8wpikjc2yj0OrBPIoOn2CjlvdA06HAgD2Ame5/A9phYPx+w4Uz/cGVjJCzfOBgl0kYC1OSXdLZ3dGsn4fHVtbfLo2nEKySHT2gg0mkfAeVZYqUznHP++VuJclLzzaHnIbzzMMHtZS/G6N7TttVZ4Ev9PdPiHpLeG0caNnF/yZIYrmnpASWgdU83nKSksBITjqgr7hbnIA/oTrBv/zax51Fh2TwaFztcsWLz/8M18OFWs8tdhOPZr5bgCqFhZaYPZPuL2j84IFvc9AVCVwsyJDwHLfqwelYfJJHxaAZvC4YU5DUs/m9T3wbnqOuE1xSGE3mtUFnKPpTD91YtVPBxhLSt3Pv6gmO7y1Tuffn8dIZYi5SVI6UjSnIFoaf28DZOm5eq94njAuhS+MVU62pgtX8xCQFIs8k0WyVjMxJdymi2I0h6ksgcgY7taXQW7px3KAm5JELCabmni/JqsBkR6B3oKqBVzhWSO9zVMeEuV5NcGdqxXs0IBKVaIlEbFCyFFkaYYlYif2hBS4JPGNbuBQO1Nj+djpZ4YmHtDjCvLsRMUbhLKroxseEFQmgGNTdf7uYAKK+46BWdLlBFzoDhOER/cltIDNlXDGVg2gqCswxfjHI/AB0QMqLbqiRZjTBHEimyLe5YW/6kyFGqPaZ/6Oa2FsLpx3pALD4UKruixd1jFuLYM/0cCv0tp2hPtKPOxuCaP4e49YEBsl9MB17D6XZvijXbEf5kMtTztv8b1dWVbjNvgb+aExKTo917zFtHOPRJSa2g9S5PmXWVeoIAtmT6vRJa7At1Y2pAtbANwrxZXnnKGTOQBwzaG5FN1YlUPronlbzxkpRptv0KEp8pRhAxYMGXanYv12t1c3fzRZr5GLSiWIEdPfyjBTFlEbamDiuL0FxJReKgNSRr1bGyY8gKi47lD3rHb9tKrPJNbpNUYNYmFM3Ygu4njADlmhgYc5iCtuT9dTB3UMKCTusEqGxjgG2jAZaNhNwvDQBwI2OMXmGT1otX6uXqjH5KotqYn3upd60mzepViVmw48qnG/6Sl+UpBJ7pnuJ6eoOyUkpM3u8ctvLl4KtKuj16gjbDwXoDM98Dhjwr/CElyRXUlQZw+4DqrEzLAiJrWIAshkdOiuKXn4j5UjhEnmN588RpC7NwqrukBYyXTolc6pDo7/JzJgzOBZdhnePZk2CQcR5/o4VgWs2rGxv4wHC+TKI2SkAuV3PBAAKhiVbFkGSWf7J5pvhl6bhEupDlFBw6gmL8UURLj7anoc58QfLNWJroYeqhfgTRz+Sj+k06g3TgUX50Qq428RgzeUtsm8rO9XPO3qiV9H8c35Lpld6MDz8wJCjyWCoAOjvg+sHu8tlKx5T622EBRjh9y3/UBEj+bf4+10Ab3mO4TqyFcGae+n0XjE6BMfQCQROrLEooVPkMIwvWaGBZ5mqJ3Bo6lvTI73Bde/eexcYFF7bhuir+vWNXi/tfFJhepDVIOj/om0jvkGxRTWymawcdS1uLiIVTMYJh+is9ykfwLUpIoCnKXzZUAhNN9ncrYqAN7MUL4GtawFK5aK836LUAyc5/DG4X+BXnKij8Nx0IxkuMgRz/5HePkU7413TRcXa/t0Q+4C/Ag5z7z+4FEPQA2VEkB/ommq76ZPciG4dyr0b5jA8UVbfhVKerwYh0b7PhfTLd7Ty+gke9zjcKbUshv/HJILaVN+sAVpt2msYp1l74wK1LPvs2w17Ktw0Ut7le6jD1yhMJkUc417115S9PXQpq390w5JPAqosjL2sb38tJOmW+psNTf4PV+VQrUbvxj2Z/R/CIioe7Y2OnOUHvWG8inm+gaWvdMkwKKqG7A2fPlZK85fAneKt2ckeHGZFmdAlmgFxKLuvsQZKIzMQzyTYyXBIgg9BT7SyyJhoFCgQtvJEmYCYgXrMthM1Ze8eih0SaR5FsjQfIB2isyyZoSenMxAbZhJQuiV5CuXBUucoOLU/GwJmNFIIro19HZx2fDoSiclDt/UWep653IKHbxyEmdP77FF9xUWu0xAsWjU7akce0kNoVZcpcHTGX9OKHgw93naHn9jDcWomXhn9iCT5mAZjB0noGIuFMVpoVlYk6JKobz+RWZa6i5A+jsQjFeoPGwAjzuHLSzxCozM0ZtTUbu+GUeftY2txQ67sBYd5yutBDWKmnWBIUxYK7IXvQV28kzssd2mladbfIKO2Kt9a0SD6cBQ0BLmpXbhXUjt6FdTWfVLulTV3QwXwYSW7nQsUIIMF3JvyEMU7jQmvBaVizld5E22MfB9/swysrZsjKwvDMTteQGlII8T76CYERwhz5eIzpl9APUzxUgq7k/Zy6IeJ23ar9VEiwiVuYyGHOKYeGhRI6omrqB83bPEeVSlKD/QTtvOx8dXP7stVg7ccDyU1VWUuXGVUl8P2FehpOgEopd7CaT6wDyV92PznRnv6axBi+r0Og2xmIxoLYk8AWgb2C+5d+O44fDZiZZipaHn9tCrSV7GwX1QnqeB0nRuh+VGbwBHmAuq7uEFGpgJc/zZNMW8VbbhONDeM0TIzlmQTI5YwQBy5AR29ScHHLzvNVecJhcTfpoKoTLI0pcpwFqrKpbpFfqTXdmy+bhUS6Iv+O8Z8XnuY1L6sJLn74wcasuDb4KMowKRzWi6KXOhnaH+acC/KO9cT7Y+VQ9TWdVDiu/JECSSR9JHfUrWAr71znIyLnaW3a6u/8sKMV2++Z0ABruFkCrR9LUu+v2gkepAvFid5LFFr/Kp4V/2kbPrtMsnuklfaPuYjkEt6gU8TnzGNAD2fnCB9AqP1dG/B3FEYIVPXxahTauSfnMIF0bn5gGdhs+CRICUbSC3c+A20jG8Vare69Svj4tmEcmLJOtVPrsJ4WAgEp4n0LbMCr7hnBsoWPJZA+kzGRV8Oy1IuuAGnXuqSov4gihjTUTENErPRBsPgIdTFlnB2O7Li/mWzWSdjfK0BcLqbxlgUw6uDaTgi9gBYo1X3Bqefevojq/BFcynpIZ1DYi4wLKu0pybSlEtQIOg2QLHRRQ6w3pVS+zaf53k7YqBf68NUrJUdXsFEWrM72dPCL402IqVg+EipdKu66LoLWhfqX6kM6dqe4WOwiZc8sw2lD+x6xpYYgALCj9K4dertD6zuR12sGOSsOJP68FWLV9+6bEwYE4l3hAD6enM0w8zCjqMK5ox4Abd5UYNPCJcU7zU6kTiBrk2iAPXgGxmTiY8C5DOW6bJukUz6X9KVQefDMXR7xrQJLot0VfZYeLlYtjTNhxYVJdNMkm8T5Cya6mF1p9eHbC4P4xPTpTlKOCsIWy7RmJgUPhxHuuZws7Z+mMB14EoDlrk3BbGS7k7tLH4JTDt6E1TQnrseXwYujJXEDfgAA2/wOpx8aBAvMYihWsSIdNXu2pR3MDI0TF5kIq5saJiJvu4BSml54wiUgyuRVg+KBG69h6kY4ZBqbfv4uxk74rbFfdMjEFR0REYVRsCVSMhc4EN3s7mu3OgM9S7SoixkeNXvLpKaE/UARbs622SWY8DKggnM+DFKqHX9xDxrdQl757X7urJjrUn1LHPkHID3+SVyCSIwsaad2sNjpsMk32k2JtV97jHufmRzKKv3TtWIpDWARIg51CDMrGuSIR34QMREMTdwrM65X5GurdA2LnWv8TFpnwzkLlAZR/OXGKJa6XfVy27heWv9SHo5P7WfsO8btk8GPXSDAyK36ZW7rQ6Zdz1WYPkORuuZ0FMksLiAW4fkamy4mSA7rwSZdRMLn7XJ9W5NXiPJTWWx3jasnktRaYG+u6UQ9ESVK8uMFNrwR17ihL18kY9gLzM89t/Wig1wQWpV+pq+tyJmL75W85uhGh/V7M6yiz0MiI3f/fH8h0mj0svx2AFi5QENsAII9tLOiwidKbhGntOwbF2D7DFNhXa9az/oPF4d/AovDcUU4WfohgskN8delV6NZMdzPQNdhaICAojlTecpRV2svthoiD/wpP9Yz3LY9fatslmvNJbjZlZafQnF1WlvKMtia6LAs/9lRwQkrYPYqE1NI7S+KL4k+uutBnpO8zS0+Y+kSrqBxn4h9+mZDyLtElcmatJw7r1XyeZTbhOjGoVH9j/8AygmMFDON1UjvQ9W5Zp4XBjTc3D20tAeuyNefq1j3ojWxFfHsH9taOmqR7DcqbiXKgNbMLWvIZPY/czxL/Kn3dy4H+988dRrVmh8+CvOs1Lwm1XcTzrC8WCvU8XkMa0hJpcXt2RHUBC8B3eidVwaTdZ2hpyZ/sSo4LqUx89vLgsCvfafhhlf4QKah/CK4XYJMfKXtxBPmBVq016E1tFplTtr42+o8QMozPL01rGZULMyqhmM62G4b5HKfWB/amLBUUrbkqg4un51Jv4Uqt2w3Bp9tN4WdvNNPGHLqc8XeLFRsVIWxo0zWJkRasVSOMQcbFNfh+PCZVWpn/F0Kocg2vyevH6DufgWccVMrnSa4LUips5+gqqn0YO/JmXZO1HfyDraTzxqEVij5QA74/pDDDfmMKVicb+Fiq9nE+UJD5HlHXmAAfZlsyvU9P+15++E2SntZwXqfIJm+nZM/519OP2HUZpQTBZhm60XBKGrfgciZjgfyA67dwedj2W8TtVkHuLfdsIjNMsir+nFQrZvx42utKMuDGl2ZlW4WlOihhZ6PdENF0+beAq9yAucjrPvMRqAXxdP/g2HuXVu/BjyRyssVu5vYFCDpW66NCLniZk8iKT/s83if3QbtrZSFK2/j7yVn59iFWcrsjzzr0dpa/9JmROYCZeOOE5oMm+1dXOObq7Yu8iKK0YtceQCg7jQV+FUeHCuR2o4k2JsP1+Gyr6m8ETmJrcCzeiAvy25oaeghBGGX0VZidAkBpZgBTRKR1Pl99S10QnEfpwBzD5ExSYjGMeMABCxPDiad8bL7pxOZM8tvmyOil7Z5YN6mmb224PAoVIHLsEBcFDL2/hrnlxajVs2E3NWFx5zhWaVTqTrk4VRTWfriRePUc35UDtFbaiJYfJ97mLZkiZvpnBOTVF5feh9SycEhA48SPMBzuXaS30gMQE8dgbG6xLv72ggCxtC/4VBr04kRA6QqN+SYTkCvOkUV737yO+OjiR7BR/KkAasqbjY3T0aM/bu/dIbm0DSKoNL72RtdsOh7WwXINbPW5WLo+xNkyIiCQtloaD1AmaEeyuqRhgTL8NAIYsaZCsUwiF/48QAfrvbHILGe/OLiuO4KBE3ZpK+fbwZhxppkmwi0pTalEttFjpM1CZyMHmVRONB8AwrJAXUT7bqXKGAALAAAAACa0c6HWBf8dAAHpY8anAQB0WIa6scRn+wIAAAAABFla')))
for name,text in CURRENT_OPPONENTS.items():
    data=text.encode()
    Path(name).write_bytes(data)
    OPPONENT_HASHES[name]=hashlib.sha256(data).hexdigest()

def verify_opponents():
    for name,digest in OPPONENT_HASHES.items():
        assert hashlib.sha256(Path(name).read_bytes()).hexdigest()==digest,name
verify_opponents()
HISTORY_JOBS=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4rYcGXFdAC2ewEYzQ272NIdWLfOf6w/HiPhVAkQ6Mq+Zhl99nEyyw0J+DioFQmFTT9Q9TmiFZA02W9Sw/4ikeph5tzRWVofq/nKuLg48fAqwnpCIhAzXmS3K9PSVxAJq3T20m3Q7vbrCNGa8On9lt3X9RZnpRLhK88MEFjhzqlETktntexG5zGkm0tvCuX/CKRLfcVbAXo+dHPkD+zM9Q3x2ucJz5N4Ynhn5Lp06rAEJFcFYYBe602VivZnCFtQlaO0Qd3lDOPnnIJ8wpP9Z1Ueo8PPjhVPsEUXaajx9qmJb/t17JmuQ4RfwnH4MO4gFWDbqMCuwmIXkPDD4TxMbWO60rUN0zaIngUJu8gws8uIZMfAWVUC6wqtpEfsrUekfWV1ruKBD9iDSi64EsYm55j0zbF3h2Oqj58Vfc88L7pGa0skLAdkXeRdA5xsUuxCNTvnePPVVww4oVsd5Ozs8B6JaY8eW6sqhX3HYt/q+hcBXEHJIG8gbKymZpLJxyaVXjq4yc2DgyoI/jd9w0uh65ew0lmYp9CCiW8dXAHvjAfMjf+9XLnHDIGlmAC1rTzyEff+hLlhG0K+XQiSlB8ZgOPH/cs51px2sJBMA3Xk5d/qTIQDKZWuLV5eq5VuqxmhHxODK8an/56JWFPVprDN5hT0zPzbuW0/m/Q+G9HENscN9kX7/5NM8Qh+IUNsNUaMShilV6phmCdpR2+ytOQnzm6Is0pzoHDgfLrKZoNVotNdNF69y0whf/MooxzM4L3IWmUwqOYpUazCdCi4SVz22CKasi7jjLk0d6iwivNV5LjeZqItros/sSUhW3Cz0jiE0uW1WWerAwB2Y3ML1Xe6LSy2PzxjsvhAzV3MBA72vNbCZ5W5nTrUPIBoDf/8dANzMYn72hJ6UWIWPvTLh+9/LvmNnEbDej6ExMMl/e+OxrbZbsQ7aG6qurtBz9v3qTlVVc1cX3znAV7MpngWjbYw792Vj/20V8+Uv/zo8sq2LU3BtCCR50uAN8SOAWraclPOCu25rDViGHFgqDZi9XVdOyQUeUdmw8Ud8Dj1rUS9NC+r2IhmQSQaS/kbsvTUfIegW+9Z/+jLtOVq7afdTt6bDBOBwfv3BlB+yO6xWk7W48iWVueUeFGZoScfgwDWe4aDBqVv83x+26W+ojvFGR2QglH/YBH53eZJ0rfFTQFWXYhcXRQ3YmbCUa0hJRFLm6S7ZfUBgdivcxi/NT/R8qzmLqhJ7ce1BXVgRPkUtWrhQc0jcD7JcaYqZ6j4Itu1s3/bUPlwEeyoymM4BxGHW2vakJQFMq6TdI8LO8oH2gHoV1glDGEqqKx1Jh/p28SprYi0RWwkWzr9OfJepxNvOBrYeq8sgLSiCDsU5edBUGiH756+wiNjvvA1RK+TanLhnYKWUXRRfQKyXKDj70p00JkiLcsTlBBhvPWr1wN1ptIrSvMzs4/Y9cec5I9SwmIZjig2U6f1AtiDc5uS4+NqqnhqxFwvmHHvZjYisGUkNDkeDB4clNT0ssNsQgBfoJeai9tznSHedZm0Lpan70chqQtBetp5W9FZE5tQwB1+ijuwweBc6hX8jeZfLthPWYTzS7aRj4u4oWODIcMZns8Hub51JZMaYDC/gvIsW3dTL+ZGdQOdYwImX3+qk2QV/anHKUOYk7Eyi6Htoc6min+iJireNB4dn9mgcKJWF8dK9oLgn1PI+O1Anb9CQnJLUWyePvCxdNCVAaTb0TTAY45DOEZDSHbAudnGoF3t8P2fLa/Ho3n3fKB3wfzmdy7hJp2Pm4I67UZWoNtKzUf5W0QQChdIYsHyaaHy9912jn5OZZ5rg0PxeXvji7hjkjHF1xav6d+1zxximCMcIUJbqvGnOGiUsfrnAciNdikFw7o5f19jzpH/uufWN/J/P5uUmGBfovdP/19CQCSrc64pZvReRuva/VQvq/9g9xF6drflWhyoPiYhvXJeg++n5MZ5TIpo120Eao3z3FjgRQFpHnSLkO8fi7xXQuw5YivqL2AcZB42AOq1hcbyJYTwdk7JB5yMO6WNI6KF/gyXukV60zPHu90DIISblYCYoDESZgx/wCYnYY53cBAgk7zQZ6fhIX2hXjob4gGvv5vBSE2odaM+P0RsFhPWIKM2XMQRPdtQGMXiJVYKvnZhUYJAP1vnF7fP9bgvjf5AwW3zhsrUGY2NjGuKdNzCwwmt/NGDOaSoBVANdBVP87tq/xl2RZEMz5Ah/DhUH5dKEXxkGm2Iy8oe2DBhpLPHEJGETjTWO0PeQFhiub1Zz7I4K7pN7O8Gq2Eoth/+1zMcOuGmo69YlI1hVdjW4dLxvQ2xdbL5Tkl7obSiSSG2a+TVk8m9nC2nhrku1BBL1m1bHV3zTjjq/mLWk0CvUxF00F7WB6K5aH/Q5VOdrCchzci3DT5foGV6U4ILhcX/TO0ozMzOwDrkKSN+pPC5Ztyg6HEZtGKrKybuLR/cYpeJTZVkqt8wLfVh70aRU1tHYL0fK9jffTMiIfkpu0Al2y76V4VkmhjwgUXWk6ue2kDWDvjnQDGunMOdpGDokP5iipI/6CC5Gw5q5KjhIBhhvR4swCAYgDGGNxT7nBlInuL5I+LXf0Ke3yc+xUXvUaes67xcRgFz3N+aGiFx/n7HO1qVd3Jsl5cqKNF0VAU/MIhZ9+pdTQtgbY8fgOh/inIaJn2af1tkEQhvDKM9ajrbydO85Dh2oyaOxkA8u4+7dFqY3U7hd8CeS4KAmPJ332gEjIq2XEU7Et3IHR642Njenxl6St0gtxQ4Fqky/ghqFacT3GZPFHgsrX+tiW4GMrcWyIcohDpkSE9WEURG64juKBfk3fnaISm06AZ7+dkQsu66E7/mBZM5yaIXrG1rxhzRDt22rlRzZolA/5XpbpactFsFT4mO3NS1zkSXOwE8XEQviCuSaLH8cMBx2C3va/Nkk3HvM+6K4Oy2WcebxZMam8a+5oSryIeO0agjW1a4i+xCHyAzSScCXnCCm5hR5R98ESTdrXiRDcsTnajAhImeHfo1VwJPyW46CSVf5rR9MaLLj2jtSXKU59Lq2Ee9cKBbXzommCcuxaIK1Bnx/uX6KPdgBk8b/TmkVAYxocgZAJwbseDKBeJV/vpuPgPCO0puq9lfJTBTnZCIKGR9ZqqZ/7DxTznpIByZA6SnKZmMDniKJDETQboumR96KuHnZiMTSCRNtypVbwvGHc7a4nTPhcFBvYMqI25saXWiZ7sPELmvTL8q4EIPycO7pQs1gxsK5OZPl3QYVP9RJoVC6x+FhrwIkGPSAzaiPWZi4w0yveemd7yPLQdVktJqBApv/PpR3wNUGFqcp7mQphWaBVNI+GjeKfztnIXnyOtS1KCyYzHoXait9C9Fie0AtFUbYV+n6IgNhCI/5k+dg1JwnRm9FfBYzO49CUNeGDhTb2djrKJbZnCLjjeKAjml47UM3K22sUmrUfxLPTbxaYy3Xhyq5BWHTk+2ns/8/cQnosqLWiM1bY3lpxPW/BZsD/PLumYKZV5J2QuVgnUwWSmzvQAyEFwrxTKQVvps2c6J6W4d6kFYbuGu5C14hjnwfimCXauZX/mPX+pyAhrPRkXXioflBqCsPaJxi5KJ+Kzo6frDGk3toH0WIdOB9K6WXrB4yoIL7d+Zho9KB5CKNAGlUo9lFk/HyzU+j/C8HcvGsR5/qF7j6cLe/kYlUVI01FYonM2qLVGnEyyBWh0cw2hHY4xDWcPRpJf1obHJjzJihcfY6QxJ7aC1m3LdAHX3UulTMH35IwxpF5rhPpSepr4SCZwUAGvQL1CUNMHzBXabcbMHVZtr7DdmGpIE+knp4D0Or3nvqzmZ5nBRQv2KaROsJhfY3c0Aj4KRFjX4FOa8NeSxJBYY6R7mwnhDO1+scf8MeiJ0zKmlvAm8Awmy4Gg+tGVcaynvuhJOsodfRNisyHo+B68+ws84Ox5ISR0gVlqfZxTtL8ThFMDKVcke0NUqUsEeFXaMrlThieig7Fy3XCDeykquSDlwICZ51LDJAYeHQzWvy4jaqmbCjsoJaca/Ct9prudEq0rZy3bfXtVhlFTuHtW6SWgOHlO9opGhTHLCmrYsQ842Dh6MY6KhwfELDX6eLQoCOd8BIqiwe8oPi6YqAe5X/7qqO6p1al66OkWGQ2+Z0AFjpZJXtryfZc9uaGDnoMEg+4nUYvkfnD+dPcGjeoWTH1GwAwHeYLdP+3g3OEoq9YowJwB2++7C7PXPwbKVoJkYW7rSWhsppEo0dGREW6HfR+nJVUymqE3ayx8tM7maub4vzKqS51iplmIIYlDBBaVe3hWTqqiMzWvg/tu6dzcgsWK71BCq5gMhZIbv/+jrAgEVFPKmz2EBE+bjhMKUO8aeUvSsZUZ1jzOP6i0FP1ihyL5D2F7Dn/Hsjl5CYFsSwCrxVcCJGgSWS132d4FvH1AZARw9TqyPNiszSY1Z3lZKxmiyntnhNaDrESU6+taeSdc61dJ6YlAa3Eow357pTR1+jLS7GX7hMSLPr4Fip785387d4fPcILCssbFHmnySxphikNDIjgqzArq3f4FeEC7XYgycQk9OOU7GMZOmmg3Y3rv0jdQuANlZA8h2/U6B3/85vgCep7N5V3yibaJXcLEtSDuEwVUX2O/rInZk9Cd5Vwneu4e2ZztZJ2DeYNf3oOOSSBfh9+7ydthhL0CC/2G9GfssqkhQYQJiXwD7ihzpOkJsfw3Bf7ust25MpOp+Mo2M3KBNy5G7G9Y7SYj+FydI2yiksxroCtSXrYVJed5b28uuuihL3xWnhCK2dENsQr3Ap4s+vsn3LaP4zaooJYNzrDhGPk0/wIrfla1bl50MC60rrmXsqxoHK2nstS//p/ebvehueeW7gSsO/QCcSlcDo9Xdg5iKPOojXsVv2y5RV5oCHd10If24NmtlhbsuwrXo+t5slbtKY2ERMVfKar5Y+Ibg8XoOVt1CKy5Hl5TMi1BMa24GcagpK96hbU+7SsQ1/11xohbBE7hvlg5+Hm14eLanNGR4YTQmqz4NqoI1YR44lIBlpx8YMrLsJNiY42SFwBWxatXqRND1A6VWVN1OJNRdd0YRN6WkwHRqGbqcvPDxlv2I+MCXnGVh/XFBm3jzLn5ST2RqaoYPHQW7cKcHMYkibWPQDxAjhzehDzrzg+zEAekd2j0Iuqde0macTSVDRKRDS/vG8p2GASLCMYM8ucHyTY8Cnqued51Ri1IBiQWBBuK/lAkhOWYHUa7irwR5M1zjXCxU3XU1Osl0enFe5hBzJj5PZSFWjSedENDYI/DtQc9vv/wdzaOFK6fQqH3j9f8TDesZuW73sngzksaOLHaptFTbIS1RmWLUQlgg9P8yrGXxWXmfW7FeHcy+2KydOUijAFpqpqHvhPpqwHCKtDDPWTfN/r7x/TbCZAy0l0ZfMRAgsdm78BIvXckajLOZuKkXvkU/eoQFJUqHBUGxWYeeQ/qGWg8D3zOPjoHGtY1+1LyZGowk3nNXTk1dKB52tf/BeZoZllg9Mh1MqIwNPZciiRfQq2ckmDnDMWEpNV/kABCK3u0+3Om7ADwXL8l8f6S/pU4+4T8n5BU+M90nLs17zk5tOYtpKMrZy8u0Guc4yVw4TFAy6ieLtcnOlY+1ZkHnXTIoLI4Eoz/0FJM6qMGb0aBRu6pvjaf6jvAqOEDyLgd50LOhUF6crJNb7R1bvduZVMI/Xgr6e8Gleq+pzygmFiXdltiSmXzr/g+MROj0w7EQhheVvQSt+dGXm4fRrdtbaWMJ1ZUSGGGaXM3jCplteEsTC+2XKk/wx2BqN7hkfdFIAk2wlBRrgi7AKjMoai79qPVvZjpOjIYYFi9DTALiXciYddd/JvVOECOzLEf2TurIlfi7bZ5xHAWCe40Y0jbKDskLs0DsVseNE28Wg/+3/uydjTgvXVadnhMvmXcz5Jk/XhKKuxvCz4jrjwwQ9v4ROVsa5FwtkUnrYLRqv9BjKdQRBX3cYO/Kf4rPuduaZo/yEGW/WMm3nOGcL/pcuKK3jqaQ66hYsQyvLGumQ1J2qt3FCLOVtCw+7ECGbng6MzyA3yiHL6wud1mRY9r7Nqy6pjOK+TS2QDQJCsK39Z7w+RgA38XFNhzEnRjoHlNq18gu9Al+1Jex5juNDU6cVtsPfxh2AhsuaxB8G6lpukobQm+5AWZXIsTLiRyNW1IqlOtzFav0L4IM9r5EWKaHaGwZXB5iJn3ElVHVwRL/5QjL+YMq+vycVSt7HLQeYildoWFWWTWfXN6w8m27SzhgRxMcVM249wf2pnQqwqfE9P1USTggi4TqoHGXE94Bl/TxoVM1Za0qKTnkcffcu219NsZs+K3NYhRlLAgnOsILTwIL8Ds46PzuuKvH6Ws4ymPmuwB19lLxR7+keJVehwQFxio9vwjt0lxPmgvDeq5bAPayPYm9IjFKZIKkrQvCOPMBfU5pWL85Zgp1uSBoZwyTkx16XsCOGWxz+tDGX7HLhxHX//MqQexUHedDuGL9M4pvrLVcaKc2GF5xPLDoA6m4qfDxPwR6Z7NIjHGmKnOXXwjYrbrODGQT/6KrfGmGxbIBI2I5tmlOzsk0UR65KkLVCMyUEzyh2HTGI0I6eFXs0D0+SevNKNmQBQmtEBNbhlMC3blUgWPbIEDMTZKCTmwxbPrQJ+YA4uW5GfoQx9IniYWWaHNfWdmJuX6RJr72/7YGaKroAA82fXGto8dh4IOCIiDtMbwofJ5XWASet5exuaxSUGfrwtm8kglJGzm+7NSVWcx2Cd00bV+5rdD+mqwirILxN9yMRBcDp01H0VClCMuZ0ycCqk3GgR6z3QDmmZkzV9vTvHiWcs9szF1opn4PmlbyPzGxLTvu0aiftvJPFgvJEbg3B2zj10Djae1hfOxleF2YUm1W9znYDcX3q/UZpAAhjQr57IHFsGSbviRVaIUUOm1NitoEljdqQigQOLUI2R8CjVGDuGbNqJgLlSub3yLjmzqkONkhobjk+oUE5bFffrLxGt84R3xmxt3agl49bKIgoDIWbOk/dEcCoBD5rL4rpJokdutua4ZwdKM9967Qsd1kbu9VWnxVmXGS1Vb9QHOCrWMkngj1OMDrtyvIcaHkOnVEFZXgcz6qz0+c1csIyCkAX6kzfVHI+nnUJoa9X4lk2NZDvD4xRsb/+KC/qIYwCPSGfoTuZITjmT2/ti2xLO4uhwy0fm4cdrk+/sGU3V583JpW3gADCk/dSgEjIPiJgJgHsCbKdeJf0quLCMpee5UjGGHriii6S24AAOzQHn8H8FPbOkNucbZXdyCvlVU2e3VbYKJBOb4jFpZRgr7rpPya0PXlAai5zRwXmDZCFRKO81N4PhT43CQx7/ZCx/hkYov2GHpIOLd9lzMtoV7MOa078AJbdoVFGdlJfpHSWUN5ziiXE37TDf+z0ww50WC/1PnxWVU5iaCwTDfYpDwcGIMS7/Bf1tTYPKE7VB/TSH3OZwp7JSP3FIADK41m5SCY18Blm9SLrq+B3maueUdRujLQVVXWS5OeWQT8JcpP9/L5oP4+miBiF33gK6RJUnFiKhMoSbcEUx9DIfEeV9gR4oDUJxNpJVwqn41udkwRM8BIqM2KhlBErvEujp3FkCynFb88aIUz2R3lM5kPKIFfer5lQqY38uS89ObPYea0pGo3Wt2Cq1o5LhKVat0v7n9oHKRbNTUySGxnhKdmpJqbkJn34+zOJgRPii45udhz6k3U1puZ0LQNoupoLrsnoubLbl0C/nH6M5RrceNfEHbvHlTIFvu52TN3/D3WibVYsqx9LWKr71cQC1Vt5sOYjhstSNTsUTiORLbR8gisL9zymBaMna4DxDkwE3baxE9LoN94fgq7R3xCmfq7Rc06Yy13Ff7rCuxcs5pGhCE83itl8CYwiORPZN/sf5OXW62jyEe1e34hMU70NXiSnZkPx9u6UBbA/m5VsgWfStKB+MIE+y7hc6Na/PIUdPYijEAAVPrh+V2YXsWiLNvuDisshjF2cyuS2OPjVNMYNdu0cMv/ORwS6aP+dSgkZ5UTixnVk1QqCptviEEIa4rpICjlqSB7KKnAlWAnBv0XGSRg88B9Pk19UvJdpA8klqcFL3P4gq8yA3ZIVY6Vi5THO738go6CyoGyTbilLibe3/w88iDI+S+S7xpyGyC7XwhIpZs9WBW5c/t+MyuPEn4Tgf3ukyN2hM1moN4mQbNGMhrIQnc4IWwQET4V2TYA9QxZahNxevVioIux9x8Yt5kOcbnv4EvW3WZ5e+jJh22ll6gjGe4NxxzbDL708ufkVed7omCL5OJQMl+WRxBqoVn1tKcaQVyf9qX/1buFJKHwENXFrnYw5AHvJ/2tf62a37pWv4FSfOHFgu0XldSAOdSSg2WkxliXgN3jXTN/Ko9GxwTl62qraLbYD3SUMLkTiNhHMep2PHT0dKjfIbAYwTqv4dToMPzxdA/4kNLJV+TZwtlSLqgNDsUSSWklMMNPHKtiou5xZaS/J6B0d8E4a7Rg+dQun7lblxOF9sPeZWftglRlNY8dd25hrM6b4ep7d7pw0Mg9KVVt+dUw/kqnM/1flG5ujoS4kr/s7TgXuD0yHnnDMAYEmfsYyJ+TjEjdw76Paf/6aJXsfcbdGTtf5kXwKlKi1EYcZFzvY8k281NsfwjkOGUPeoaWjsdROHZQvsRBO49C30snBTwAAAACodT94d+U3QQABjTOd7AoAuunHkrHEZ/sCAAAAAARZWg==')))
ABLATION_JOBS=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4DPsA2ldAC2ewEYzQ272NIdWLfOf6w/JCHIV9LYEBrZx/yia3YEy7NPXxLo1HJsR4PxZiHryudIBftm191Evab1JHF/AhKbi2P3veb7UjBZXOdKnji3MSJiu4MlA0A/6LynGDraRIJIuPlkLhR8H+tU4q1ajJhRW9t5QN2Emx0wXxaO6ZiF+VZC7LwstLuxkndKJFJvBOE+zfrwNQZylq5VcTLEcNQq2UuQWyIVoinuQtrcN+GzFo8qvxb+AqNIXdwgXzk9kEzRs+R2k6oGVxFsd9pi0OlBsZBzsPi7QtBDuWZfnCLKj3xhN08Duf54t1Snecw8GwIPaUM5qfTA61rYAVNEqQWmeCXLkqmdIcbdzvpqQdzh9QDnlZamjC00dqtDt2h4h1M0GRPGKVhgEKQ+ebAf142z5uSHCDTNxL8IEkdhyXPABOLFgjnOyIq7soCUYmjpibkz+8glAnOCZRt+JGAqaPJvxNqtIC8LNihcdTKF3oMMjR5o2eRAgfXJSA9bDsDy0HyHVWyNVMc+uM1SppoKsTi5gWcMubmiC2FazGXxmoXXLLseN2U/JXRDMyytE4u5Wwodywi2wL3mfF3/6fZmYakjInrim+SfiavrI8ycot9qMX13uRMxUZBZoYym7Ex5EXmZRqZ8TDJTgq92IdN/ToWxNVbl3LGXvbfanT9SdVkJxS7bTsgjvTjPhpe7nrHvSl/sJk23I+mnYx/nULqe1y0nOKrrBSAEbnPxWcJM2drnNymp3k/0aat2LWzRi5EC7SsGrXZ1Q0TmlVlvX3pQJhvFiDJXd6IkvFF2QjBcKSrxYaO04ElmLt6OITUosjIQmc8PMrzVo1vfGAzunCp6ZNp7GilisAay0LGyk9QdO4I5A8ZD7/YezKA7H1WPSgs+/yEw955iMo8Fq+rcDVnINSmyJr7/a11rvaeJNhydLugt9Id+0XVSEsA4Dd34JykPirBqTxkATwGXXDk7EGKRqVF7xfsuUeDjP2hV7oiTmz55kc923vQcobQn4Mmk40R0bqbFAv20pCJSgp2GivJ/4sy9U/c4VP7bMqFSNiDoY/puGbLwbRY0PIHhUYRWW3gLspfpkRLJ2UrVr8SoKewQbPJZXFJ6hFnBbqVyOVeIFQPF169U2QzxKWEmI8vLkPPKCcf/yMIi/ZtJuAAAAAABVPQvKmab37gABhQftZwAA5smFVrHEZ/sCAAAAAARZWg==')))
REACTING_JOBS=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4HWgA2JdAC2ewEYzQ272NIdWLfOf6w/HiPhVAkQ6Mq+Zhl99nEyyw0J+DioFQmFTT9Q9TmiFZA02W9Sw/4ikeph5tzRziq0ULsCYcmUYYSjO/ohq7mYFZwkc8uPxQ5XrAYaumk9Fi5ZUOgx8qd7vHdNpFzkyTHiRurZsoaxgM8VOcN94kvLDZwitgtUeULNTOy128nOU5r1kBsHA4wgSFAmbAgf0Dm7+QaGi6+/jf3M+pf+idyHMllomaWilO6pkHRTSjub3uX148zRcRe66OduFznpgzz0sj89ZYFGjoHa7OYOkSv+xa2q6StSFgGW+xpe3DWzE+j4VicRwT2PWDI1FiuIlE8gXf1RGoAhImVGsHgwkmyfYuYQx8aCXHnoGWImMwsKeAmwzTlRRC6yZnMMN9GVbaHMgzFHIznl0jTknYHLFBM8tOPBJUptMdmnfofT4gDmUY6+Zk3iNTeqwISpfogWOU/IOtt9x1vv1aJxBeZ+WrHyMJiSqdlPV929pCp3oTEGxYlPrKR18JJ6q6JNNPnwkBvWF+//Rms0ezhmwDsKGVC743cS+T7zfawf3xeDhXWVDBaj0noYtJhK5vjfEsTb/ZQNYZMH3LFP5cArxP9QRjahaGmmTpCtzf/ZzRfhBtWAMszRK4+uFHG1nZpHyLg37n1EBYxVz+xUBqJm0beQf73MoYlERJikkFrDvd/kC7vQPilddRO3Xm1sPAOp2AGsvyEXOf95xfX+pGOMbieU5Jr2GqIsf9I8q1p7yGe1rIbRTlgL3OPtu7IdGPAmp4mVI1oH/psx5tgqJR9gmzU52w75kAtUbupLks/AHvRogFrtK4Doqjf9FeJONjpbcDFaI6844A1KuCe+oHfk3+ASPjo1pIChqALocgyi1yF4FAIfci+DiT9kLk8gffDJvVPwfKrS5znRmug2XdEaVqsIvZuB/Th9HxoQdi2KvoPBnzm05FgIX0nvWTv80MLvPMUblsv0NhUirxtG1n4PpHuyfnDKMqccyYhSDeBXORs4AFHoWtFMjHvr+fFGVAD/n3oNCL2fh1Amh4Kqc6F17+4ed+I4BeL+SS/vppbLYX1E9zA90GhAt8mN9nIPW/yc+ROtGym+FG44oxxoodXkwCaYXGEgkTzJvuazT7GHPKnwjJkZZgN1UAAAA846m1mU8kqgAAf4GoesBAHQp9iqxxGf7AgAAAAAEWVo=')))
DEMO_JOBS=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4AaoASldAC2ewEYzQ272NIdWLfOf6w/IsRVbV9sGylrAJsaxeksgwQV1J8Ugl7oXXlx+MecYscfyJSploa3fEERt5KAsvGmzFaRSE+eTLHYXMcxxg1vYc7/ODNa6TaT/TZzNprp23r+xRQYvnkadgalcXH1xNOahBZkb2z6+TIfUQp4t5/uD2f+g5TnWfgK5N2aT3BpepKcAeK4oHLnLUSuGWjyvxS9vQKphtDF0sbgGVky06sJBEgesTHqlVcBb/CP0b2M64pLqUPXpouF0uIkBrlfeJHBT7vsZDt+FYDnbQ4UORHOoy+IKMDbU+lK/wQlQGMpI6cu8O9XG5Fbugz4sVBeiJiK1bUKKD/xxyysTcVEkOw00ir2E6pecMkYVsl+ewc9HrFPaAkekbZeuVgAAAADa1Pebzf/cJAABxQKpDQAAZs0LwLHEZ/sCAAAAAARZWg==')))
Path('episode_worker.py').write_text("import contextlib, hashlib, io, json, sys\nfrom concurrent.futures import ProcessPoolExecutor\nfrom pathlib import Path\n\ndef play(job):\n    with contextlib.redirect_stdout(io.StringIO()), contextlib.redirect_stderr(io.StringIO()):\n        import kaggle_environments\n        from kaggle_environments import make\n        from kaggle_environments.agent import Agent, get_last_callable\n    assert kaggle_environments.__version__ == '1.32.7'\n    source=Path(job['candidate']).resolve()\n    rival=Path(job['opponent']).resolve()\n    assert hashlib.sha256(source.read_bytes()).hexdigest()==job['sha256']\n    selected=get_last_callable(source.read_text(encoding='utf-8'),path=str(source))\n    assert selected is selected.__globals__.get('agent'), 'Kaggle selected the wrong export'\n    originals=Agent.act\n    calls={str(source):0,str(rival):0}\n    errors=[]\n    max_time={str(source):0.,str(rival):0.}\n    def tracked(self,obs):\n        action,log=originals(self,obs)\n        key=str(Path(self.raw).resolve())\n        calls[key]+=1\n        max_time[key]=max(max_time[key],float(log.get('duration',0)))\n        if isinstance(action,Exception) or not isinstance(action,dict) or 'Traceback (most recent call last)' in log.get('stderr',''):\n            errors.append({'step':calls[key],'agent':key,'stderr':log.get('stderr','')[:1000]})\n        return action,log\n    Agent.act=tracked\n    try:\n        agents=[str(source),str(rival)]\n        if job['seat']==1:agents.reverse()\n        with contextlib.redirect_stdout(io.StringIO()),contextlib.redirect_stderr(io.StringIO()):\n            env=make('kaggriculture',configuration={'episodeSteps':720,'seed':job['seed']},debug=False)\n            env.run(agents)\n        assert len(env.steps)==720\n        assert all(str(s.status)=='DONE' for s in env.steps[-1])\n        assert not errors and all(n==719 for n in calls.values()), (calls,errors)\n        seat=job['seat']\n        own=float(env.steps[-1][seat].reward)\n        opponent=float(env.steps[-1][1-seat].reward)\n        if 'expected_own' in job:\n            assert (own,opponent)==(job['expected_own'],job['expected_opponent'])\n        timeline=[]\n        for step in []:\n            state=env.steps[step]\n            farm=state[seat].observation.farms[seat]\n            action=env.steps[step+1][seat].action if step<719 else None\n            timeline.append({'step':step,'target_tile':farm.tiles[4][2],\n                             'hand_3':action.get('hands',[])[2] if action and len(action.get('hands',[]))>2 else None})\n        return {'seat':seat,'seed':job['seed'],'reward':own,'opponent_reward':opponent,\n                'margin':own-opponent,'states':720,'candidate_calls':calls[str(source)],\n                'opponent_calls':calls[str(rival)],'runtime_errors':len(errors),\n                'max_candidate_seconds':max_time[str(source)],'timeline':timeline}\n    finally:\n        Agent.act=originals\n\nif __name__=='__main__':\n    jobs=json.loads(Path(sys.argv[1]).read_text())\n    with ProcessPoolExecutor(max_workers=8,max_tasks_per_child=1) as pool:\n        rows=list(pool.map(play,jobs))\n    Path(sys.argv[2]).write_text(json.dumps(rows,indent=2)+'\\n')\n", newline='\n')
def run_jobs(original_jobs,name):
    verify_opponents();jobs=copy.deepcopy(original_jobs)
    for job in jobs:
        actual=hashlib.sha256(Path(job['candidate']).read_bytes()).hexdigest()
        if actual!=job['sha256']:
            job['sha256']=actual;job.pop('expected_own',None);job.pop('expected_opponent',None)
    Path(name+'_jobs.json').write_text(json.dumps(jobs,indent=2), newline='\n')
    run=subprocess.run([sys.executable,'episode_worker.py',name+'_jobs.json',name+'_results.json'],capture_output=True,text=True,timeout=14400)
    Path(name+'_worker.log').write_text(run.stdout+run.stderr, newline='\n')
    assert run.returncode==0,run.stderr[-5000:]
    verify_opponents();results=json.loads(Path(name+'_results.json').read_text())
    assert len(results)==len(jobs)
    for row,job in zip(results,jobs):
        row.update(candidate_label=job['candidate_label'],opponent_label=job['opponent_label'],tested_source_sha256=job['sha256'])
    display(pd.DataFrame(results).drop(columns=['timeline']))
    return results
```

## 7. Four grain, 27 milk, still a loss

One guarded route drawn from a current rank-five public episode exposed a physical failure. At callbacks 171 and 195, our workers requested two wheat units from an empty shed and received none. Four subsequent feed commands had no wheat available; a cow eventually escaped after two unfed days. Eight cows were purchased, but only 218 milk were produced, harvested and sold.

The **V226A feed repair**, now included in the default agent, forecasts the next wheat pickup using our own existing worker positions, pending commands and current plan. It buys only the full shortage, bounded to four units per turn and eight per day, with cash and shed-room checks. It skips dawn, route boundaries, conflicting purchases and the final day. Worker commands and routes stay unchanged. It uses no rival identity, hidden seed, rival private stock or future observations.

In both diagnosed seats, buying four grain across two turns removed the missing pickups and failed feeds. The cow survived; **245 milk were produced, harvested and sold**, with no milk discarded. The margin improved from **−11,816 to −2,566**, so both games remained losses. The improvement was not solely milk revenue: our reward rose 6,735, the rival's fell 2,515, and our milk revenue rose 5,137. Every changed cashflow is shown below.

The eight discovery games had **two wins and six losses both before and after**. Six zero-purchase controls reproduced both rewards and terminal states exactly. This is a verified repair of one failure mechanism, not a demonstrated public-score improvement. An independent editable copy remains in `feed_experiment.py`. The dated 474-cell transfer comparison below is retained for reproducibility. Section 9 adds the completed fresh comparison and later public histories.

### A win caused by a larger rival decline

Public-history control 107191256 changed from −1,158 to +3,018 in both seats. It also bought four extra grain and sold 27 more milk. Yet **our milk revenue fell 1,231 and our final reward fell 912**. The rival sold exactly the same quantities as before, but its final reward fell 5,088. Shared-market repricing explains the improved relative outcome; describing it as a higher-income farming strategy would be wrong. The complete executed-unit and cash deltas are in the downloadable report. This case has sales accounting, not the full animal-production trace used for the rank-five diagnosis.

## 8. Why the supply predictors were rejected

Correctly reconstructing past market flow does not mean predicting the next sale. We tried using positive rival supply from 24 callbacks earlier. Across 45,277 scored predictions, precision was 28.06% and recall 13.99%. Accuracy looked high at 96.38%, but predicting no supply event every time scored 97.03%. Another 3,023 ambiguous predictions were left unscored.

We also tracked publicly visible harvests and the worker's later return to the shed. On 21 already observed histories, this potential-delivery observer reached 42.90% precision and 23.82% recall. Its 96.44% accuracy again fell below the always-no-event baseline of 96.70%. It cannot observe private cargo directly.

A candidate that used this observer to gate the timing rule gained two wins and lost two in 42 discovery games. Both agents won 31 games, but the candidate's mean margin fell from 5,240.98 to 3,885.83 and its worst margin fell from −2,890 to −3,495. We rejected it as a replacement. These results do not show that every possible delivery predictor will fail; they show that these particular signals did not support this particular change.

One further trap is duplicate opponents. Dynamic Route Agent contained exactly the same agent and plans as Aurax v2. Most Powerfull Route duplicated the SevenTurnWide runtime and settings; only JSON whitespace differed. The additional Evgen chassis has different reactive code but all 13 plans match Shop0909. Different notebook titles should not inflate the apparent diversity of a benchmark.

For your next experiment, record actual purchases, pickups, production, sales and both final rewards. Separate an avoided loss of goods from a newly won game, keep the regressions, and identify which opponent implementations share their production plans.


## 9. Why the feed repair became the baseline

The four-grain repair passed the retained comparisons without losing a parent win in any panel. On 318 selected public cells it gains four mirrored wins over the timing baseline. These are two histories, not four independent discoveries. Later public batches are included; no unfavorable case was removed.

The new reacting comparison was 48–0 for both agents with identical rewards and terminal states. The feed rule never activated there. That is useful regression evidence and provides no active confirmation of the repair. The source review excluded notebooks that duplicated existing implementations under different titles.

Separate crop research also produced a warning: a carrot controller grew and sold all 150 additional carrots, yet worsened a newly collected loss from 5,936 to 8,978. Another selected history improved from a 3,545 loss to a 2,571 win while our own cash fell. The crop controller remains experimental; higher production and a perfect selected record cannot justify promotion by themselves.

## 10. Audit a loss before changing the policy

In public episode **107426389**, Soil's submitted carrot agent lost by **4,253**. No added carrot plot, tomato plot, or emergency grain purchase activated. Moon's baseline matches all **719** recorded actions, so this is shared-policy evidence.

The rival sold **420 milk** to our **245**, earning **35,960 more from milk**, plus **9,087 from tomatoes**. Our higher strawberry, wheat, and wool receipts and lower expenses did not cover that gap. This supports investigating productive capacity; it does not prove that copying the rival's twenty-cow schedule would work.

Other public losses look different. In episode **107436133**, both farms sold 161 wool, 249 strawberries and 245 milk, yet our sale proceeds were lower. Episode **107435114** combines a wool quantity deficit (253 versus 281 sold) with a carrot deficit (84 versus 155). A single “buy more animals” rule would not explain every loss.

The table and charts below use executed trades, including capital and hiring costs. Both cash balances were checked against the next public observation on every turn. The accounting method also matched every event in an independently recorded full-engine ledger for a separate episode. These remain descriptive accounts, not counterfactual estimates.


```python
PUBLIC_CASH_EXAMPLE=json.loads(lzma.decompress(base64.b64decode('/Td6WFoAAATm1rRGAgAhARYAAAB0L+Wj4H66DEBdAD2IiKdDz8pZdnKPE+v4dULWa5C6BKpao5V4o6gxNUbi/jdZIgV5fP+YeMspNNI7Lc/0klKv5CNoklx+WD4tA1ELVjESFK81UnbR7RfdI5xdSFOhtOyAoWWF1uL4oKRwGUBIABUvlwUO/EVwTwQZEUpry5x7m3s6cxJy6qpgwxEBy+GQZp1xSM7Z6NaW4KLw6F4FfoFALmVlCzLfBeghX450DD0/FcSrTrx9M6e1iIFApLxsV4lv75bev6VqSaJ879LIuxXxRE0MhePD3P/n1lFkZIkz1wg6HC2QHQVI0kSNM6dAPQFJC6vRYEnPF1lnXI1sMk6B+12YeeW06hH7KqJ1UkKbGwk+14t6JoKonI1xxkQb8mvmOA10T/3N/uveXjLTzatWxwGzEam+0LiIMCQ6oX9Nbry5+0U2aWw05HOqCsSzuVvAjLe07r+G8zrYGJjmDE5IUiunAm9NUeVJNpMkm/neNjwa1iY5nukk9BV74OyC6pw1JBSLnHAtYQBRTP6eOswnj6wZOOvCUCgnjXXrGcoUVPdsPPFoSBh6a75M6R73GzVLj8rAM7I0+N0bHq2AWunlhf4aaafqLt0oLcpXjpZO3xlM3gXE38co6Xwrmc2MoL5daBHDzk3H7zORzTRLWEM/n4zP8+SW/6trFVxjlc8nLgNm3iUvsqfeoqIVv5gaEr6Y1cZKVkGl1oeXnI5BWhCTHBymkxEuFlX+5M6DfBTKVHUwTgicGfbsXet47B3uoURKRi0vjMpKZC08ARUKuyAATXvOosmS5jF0GTKfCiGMRZT/W2/Sqs3BkvkH20Zot1CnKusa4yCNg4+/uF+f4mCSpv2ecBK5l/idXLWyzNYuA7r3VBSg4Lj6GB+J8B4hJQmu2eOO3fRohveNSC1+zI1WPE8oh9kFPDDpfhvG07MQGqGue5IqMATOWtVjn8ggc+x7S5zFdaALUnb64oCT5jNM8IY+doZeoz6tf6bwQKbT5FxkhYY8Ay8P8yO4vAOP5YFbahvBC5ZJXOBW3xqp8w75kHKb8NekT5LMgU2M27xbilRmgFddKDMb9Bs4sJ6PYLvqLpntKjJniXrRnCE0ofGFjzdNaQMo0rjpluQaotSJ51vg4m7hqd312dHETVkaCRVss7Hv8VQBWfwIynV3WNaKEcsRaTlww76MG78goAXC/U3AZCEvCMd90KOSZGtgZyqQHEOnVPRhvDs8hBoQBur6fhRUweVpU+5fuY8B9DisueG6PK9UoP01nNjQEpM2HZwEsrwokIfcaJl/WM0EyK9/ZNLjJgAoPOiOoROZAaGxyMvkc/I7bm1FS0uwE7MIM3kTL5ERztSK6bYknsigQ+6WtocGFYfFNODhz8J8hv2tBIGZ9l5pJF8sY0FVoe5PYh3A4gXakqnWEpGXa4A6aD5QONTQZSQpjJs57NDBgDRGhfhai3ckxhKTDljm86pYK3j1NX9nbUt6DBR9R4YnSHDaFFH9VcJbrnKn9FSGBTTqsna/GMYOQSHTQPc9C8wZwz2QgVg1Hxv1MCNSb1JTMJYVXZcgDKAP4Ag5Qb9UCIeLHbMD7cltr+04+wqMLqOGTYjjKeA1zERjb3piWZUqaYUOZCdVZNSJAvlHR8UlQobXWmxGMJ/FetSHPR5Lgu1D+xGCD6dWbROHjSM0LzZCVq0ITAhyxsH8vkaLVqEzDBwVaciTKv7dvqfFo7j16mwiixJqEnduStYl41lskEMWaaNE/p0/53U7FoRtmd3lG7VHvF92+AYL5BIw4dqcynhcUPSzIzKTx5S/7BcwisdViKqCyK0tOuyJ++uLNq4D9xDJnCYJjrg7wiIhQwc2DnJGP0i7TTvBmQ8C0qDAQYm4fGFWetrS7Ft1pAzPDDwOUn/Sd+wWuS7oSgjHAwFiQarIqvZO+DxAWkNizykcBlA8nRfNhe/a9xGu2cin12JJsmHRBj/7AA0eX8JM4AuN4dMHbPfI4VJOC9xLdjEux9Vb6WalPVtruF8/VxFxFX2bOKKAkhmq8gZM+QqyPYpnMCJg/jptFSMVJ/ccb8G+4jBuqtxX9/nUYD2rWPsrwsWSb4fhLOvu2AJcSZzcxbd+J82JdgkKDypcV1LMMaJ319KdBGSlBHk8bvu/eestF6HFVjQFBgoypK/AD7hhOobNL619hXqa8LBGgHQdyK+2ItJwi8PDDsEg3IfA07SsHgFnRJF/Lrvfq9sWdXDl3Znw0iQegsxXx3LfPYmDCq08G+NXNf82S1rFhPa38EMzFyqwhBz5RUQFtGIoB0cmCUVAU/NYOwHJVz5fvzfvNnQFuqzbzLC+XBZz4pkpLDiouEnvR35M+OJ7qHwggEsgVCDeyVXnvXt8OedsUYRrvTbpWFgS/VHcnEnaOvMGXfJCqZ11BF9FakrcL9pj2eYWSzTTE3K1UbuyXdQZBH5jJniG9jaU4Q0kbEM2n+VaDf9wSOpbmO6ZeU8XqgTBx7b+hgCQZMb+WyO3r27+GvDXrzQEX4mACmQ5J1D58r29M2WzPx4sJou7EhJw42S4clxBjHYiZkc7HLDd9syNtVoeEjlm86S8Up4bWj2R+pXXcMXEh869jAt/dd83S4Yx5Tw7uqieBCpkj6b1nTSTdzmMMVlqW6UTBt5mVeyoXKka/f0qz/WbJ0+v9a7eu287J2MzFJ5Tix6CCiFXiqTCnwqvQrhItmKfn6F9VtDp4i5x00cXd+2Npy70wAF1XAqHVImjo7JO9uvKJhv9CZ2IPicZlT8R41eVIF86a9BqkOwutlpuBhuRsb92Z8JkQ5LmvyYuC/RYKj2A8VJ+lxzJ1OFWcywamE03pzSaZQDxUT6J38PuGUEFUBk2htL+XQy+HrpUiUI3OcwyBiV7T8yl4T685b6IxyzSOGZZ4rrGxY+SxLKxOWzldQQlSIMJs2VqspiRmTtHI38wlV/TD1mnBNb0M0xTB0Sgg8bmG4+Q0RxvWNFSP35aPAGKhs6+iaovwN07Rqfu2eMmhnem8UNYBqcbbo1CwMURy+IkWoehHc9DNhsH5mA5MIE0oz6gcSMzjr75kHuxR1NsXAgpMXE4nR5xkBtkKh8551afy17G/BXB0NAQD/4ZA2lnlmOkxEKsK2p4t/auw7VawCIIJTC040zGriX8oDDp4dvcbjiYCKOFQ8tr2W9uA4KwoIh8sF5QZ5aXtTc8Es5joqR7+S7RyiinMuHrVv5htLwDgymkAEeuk/p4dnibXWOMcHjFHqJi31GrQ748qLBHw8mddM5fBH3BlJQ13sn7lJOPy4UhPxRx4+wZix/p7ynXTf4peOBk201oCxJszrcMzUomQCZUDB08xTzTc6Ncqk0eXwlwxwRZLPL/LEzHsAf/5uDVO/UMDfdv4soMUxNd1FaAIOTMeuAHS9r6P/Xa3UGm3XfFzhnxhFgOfOZsAtmi/F0bP8ddGo3HqKp2eoRv4xumOnXj58yosuIpGHfOlfxKZUoW7oU7XuwRhi1GYV1khot3cjmrWJjXzclM5n5ZLkL1JWhGGYI9ymZpz5+/4B2zhHyEGe9MvDXyMsIVRyxeu4gPqVDIfH+Xyo2uA5VUE3RGU0Q1WtYqkOLInoh/5e3rFK1XTFKLZ/ekfCWRUfC9iYHLs93P6yGm2jtgQzHQ7XKfq1m4WIq/r9bdjysyTj4orp9u53HzdV8B9BndYCltMJRgsfsTRKXcZGRFlMKUzC8qmw0EaQ1Uss3zmQzcIywOBmyR2K/xJFGzg/nXFEygXabTQJtXwhybJ3/dA1zK1GtC9Az7RdPVosEwcmkq2ELEgaR+cHLmx9A3n9MS8irr2aN1DEtNr6ticXQSa5m9PyhpXHj/VjahPD8yu2ZSIksyxVP1yARosV3X0OkwcboxGEHt91mUmTQVm4PWgSVKZg7R1pgrXDrA2m3CJ45+0EGLJFbhKWGH+rWYoaiWCB/AjhXLJBwRzA4JDqjkRbcav5vdu4eT7Qr98VxGxRg9YJmK9llOroq9zZYN4LkGV76X3XyNp0aOdTkp6JNs0QRGX7OBxUNSR6wYWYOnfXvbvB7gt2pprYd+b8/N/1F8XoSRwnc6BbSI8fp3hvzO8UFs6127J/k9stA/pD3JZsEwdz7hVlUVxjxXevxAYd2G0hAK17tAdILBBYIf+HYSBv/o8ULs6vR9y4ABPmBXYqFVkx5nPo2BTf8aWyquHusA/Tdeda5ddyMAAdwYu/0BAJPMTZixxGf7AgAAAAAEWVo=')))

def cash_gap_table(ledger,own_seat=0):
    own,rival=ledger['players'][own_seat],ledger['players'][1-own_seat]
    rows=[]
    for name in sorted(set(own['cashflows'])|set(rival['cashflows'])):
        a,b=own['cashflows'].get(name,0),rival['cashflows'].get(name,0)
        ua,ub=own['units'].get(name,0),rival['units'].get(name,0)
        rows.append({'operation':name,'own_units':ua,'rival_units':ub,'own_cash':a,'rival_cash':b,
            'cash_gap':a-b,'own_average_price':abs(a)/ua if ua else None,
            'rival_average_price':abs(b)/ub if ub else None})
    table=pd.DataFrame(rows)
    assert table.cash_gap.sum()==ledger['recorded_rewards'][own_seat]-ledger['recorded_rewards'][1-own_seat]
    return table.sort_values('cash_gap')

cash_table=cash_gap_table(PUBLIC_CASH_EXAMPLE)
display(cash_table.round(2))
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,2,figsize=(14,6),layout='constrained')
largest=cash_table.loc[cash_table.cash_gap.abs().nlargest(10).index].sort_values('cash_gap')
axes[0].barh(largest.operation,largest.cash_gap,color=['#b35c37' if x<0 else '#247f78' for x in largest.cash_gap])
axes[0].axvline(0,color='#666666',linewidth=.8)
axes[0].set(title='Largest contributions to the final cash gap',xlabel='Our receipts minus rival receipts, including costs')
trajectory=pd.DataFrame(PUBLIC_CASH_EXAMPLE['cash_timeline'])
axes[1].plot(trajectory.step,trajectory.own,label='Our farm',color='#247f78')
axes[1].plot(trajectory.step,trajectory.rival,label='Rival farm',color='#b35c37')
axes[1].set(title='An early lead can reverse',xlabel='Completed turns',ylabel='Cash')
axes[1].legend();plt.show()
Path('public_cash_example.json').write_text(json.dumps(PUBLIC_CASH_EXAMPLE,indent=2))
```

## 11. Editable cattle discovery arms

These are **unsubmitted experimental policies**, not a replacement for the default agent or a claimed improvement. They test whether changing up to two or four scheduled sheep purchases to cows can help when the currently visible shops favor milk. Purchases, pickup, placement, harvest accounting, and sale capacity form one coherent change.

The gate requires at least two currently unlocked milk-consuming shops, no Yarn Store, milk price at least wool price, and a compatible existing herd. Purchases must be confirmed before workers are redirected. Travel, pasture construction, care, feeding, harvest visits, and hiring stay on the base schedule. Extra milk is sold only at an existing sale slot, bounded by tracked extra harvest and available stock.

Mechanical checks passed on a day-eight segment: both variants bought, picked up, and placed two cows, handled failed and partial purchases, and respected extra-sale limits. The cap is an upper bound; unresolved animal delivery can prevent later substitutions. Full-game evidence and broad confirmation are separate requirements. Historical livestock experiments on an older base do not transfer their win records to these agents.

Run the next cell to write `cattle_cap_2.py` and `cattle_cap_4.py`. Edit either file, then enable the final benchmark cell. The 24 jobs compare both arms with the unchanged feed baseline on four selected public histories in both seats. The supplied rival action schedules are diagnostic controls, not recovered adaptive opponents. This small discovery set cannot establish a population win rate.
